```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image =cv2.imread('/content/Ustc.jfif')
image =cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
```

```python
h,w = image.shape[:2]
```

```python
tx=100
ty=50
```

```python
M=np.float32([
    [1,0,tx],
    [0,1,ty]
])
```

```python
translated=cv2.warpAffine(image,M,(w,h))
```

```python
sx=1.5
sy=2.5
```

```python
M=np.float32([
    [sx,0,0],
    [0,sy,0]
])
```

```python
scaled=cv2.warpAffine(image,M,(int(w*sx),int(h*sy)))
```

```python
#center image
center=(w//2,h//2)
```

```python
angle=90
```

```python
M=cv2.getRotationMatrix2D(center,angle,1.0)
```

```python
rotated=cv2.warpAffine(image,M,(w,h))
```

```python
reflected_x=cv2.flip(image,0) #x axis
reflected_y=cv2.flip(image,1) #y axis
```

```python
plt.figure(figsize=(8, 25))

# Original
plt.subplot(7, 1, 1)
plt.imshow(image)
plt.title("Original")
plt.axis("off")

# Translated
plt.subplot(7, 1, 2)
plt.imshow(translated)
plt.title("Translated")
plt.axis("off")

# Scaled
plt.subplot(7, 1, 3)
plt.imshow(scaled)
plt.title("Scaled")
plt.axis("off")

# Rotated
plt.subplot(7, 1, 4)
plt.imshow(rotated)
plt.title("Rotated 90°")
plt.axis("off")


# X Reflection
plt.subplot(7, 1, 5)
plt.imshow(reflected_x)
plt.title("X Reflection")
plt.axis("off")

# Y Reflection
plt.subplot(7, 1, 6)
plt.imshow(reflected_y)
plt.title("Y Reflection")
plt.axis("off")

plt.show()
```
